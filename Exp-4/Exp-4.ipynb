```python
import pandas as pd

# Load the dataset
df = pd.read_csv('/content/Global_Superstore.csv', encoding='latin1')

# Basic inspection
print(df.head())
print(df.tail())
print(df.info())
print(df.describe())
print(df.columns)
```

```python
categorical_cols = [
    'Ship Mode',
    'Segment',
    'Market',
    'Region',
    'Category',
    'Sub-Category',
    'Order Priority'
]

for col in categorical_cols:
    unique_vals = df[col].unique()
    print(f"\nUnique values in '{col}': {unique_vals}")
```

```python
print("\nValue counts in 'Category':")

print(df['Category'].value_counts())
```

```python
print("\nValue counts in 'Segment':")

print(df['Segment'].value_counts())
```

```python
cols = [
    'Sales',
    'Quantity',
    'Discount',
    'Profit'
]

for col in cols:
    print(f"\nStatistics for '{col}':")

    print(f"Mean: {df[col].mean():.2f}")

    print(f"Median: {df[col].median():.2f}")

    print(f"Mode: {df[col].mode().values}")

    print(f"Range: {df[col].max() - df[col].min():.2f}")

    print(f"Variance: {df[col].var():.2f}")

    print(f"Standard Deviation: {df[col].std():.2f}")
```

```python
high_sales = df[df['Sales'] > 1000]

print("\nRecords with Sales greater than 1000:")
print(high_sales.head())
```

```python
technology = df[df['Category'] == 'Technology']

print("\nTechnology Category:")
print(technology.head())
```

```python
high_profit = df[df['Profit'] > 500]

print("\nOrders with Profit greater than 500:")
print(high_profit.head())
```
